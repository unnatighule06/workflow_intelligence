```python
import pandas as pd
import numpy as np

print("Workflow Intelligence - Data Generation Started")
```

```python
import random
from datetime import datetime, timedelta
```

```python
np.random.seed(42)
random.seed(42)

num_records = 5000

print("settings configured")
```

```python
departments = [
    "HR",
    "Finance",
    "IT",
    "Operations",
    "Sales",
    "Customer Support",
    "Procurement"
]

processes = [
    "Employee Request",
    "Purchase Approval",
    "IT Support",
    "Customer Complaint",
    "Leave Approval",
    "Invoice Processing",
    "Recruitment",
    "Document Verification"
]

task_types = [
    "Approval",
    "Verification",
    "Data Entry",
    "Review",
    "Support",
    "Processing"
]

priorities = ["Low", "Medium", "High", "Critical"]

teams = [
    "Team A",
    "Team B",
    "Team C",
    "Team D",
    "Team E"
]

statuses = [
    "Completed",
    "Pending",
    "Delayed",
    "Rejected"
]

print("Workflow categories created")
```

```python
 data = []

start_date = datetime(2026, 1, 1)

for i in range(num_records):

    created_at = start_date + timedelta(
        minutes=random.randint(0, 60 * 24 * 180)
    )

    processing_hours = max(1, 
np.random.normal(24, 12))
    waiting_hours = max(0,
np.random.normal(10, 8))

    completed_at = created_at + timedelta(
        hours=processing_hours + waiting_hours
    )

    data.append({
        "request_id": f"REQ{i+1:05d}",
        "department":
    random.choice(departments),
        "process": random.choice(processes),
        "task_type":
    random.choice(task_types),
        "priority":
    random.choice(priorities),
        "assigned_team":
    random.choice(teams),
        "created_at": created_at,
        "completed_at": completed_at,
        "status": random.choices(
            statuses,
            weights=[70, 12, 10, 8]
        )[0],
        "approval_count": random.randint(0, 4),
        "rework_count": random.choices(
            [0, 1, 2, 3],
            weights=[65, 20,10,5]
        )[0],
        "waiting_hours":
    round(waiting_hours, 2),
           "processing_hours":
    round(processing_hours, 2),
           "sla_hours":random.choice([12, 24, 48, 72, 120])
    })
print("Records generated:", len(data)) 

        
```

```python
df = pd.DataFrame(data)

print("Dataset created successfully!")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
```

```python
df["total_time_hours"] = (
    df["waiting_hours"] +
df["processing_hours"]
)

print("Total workflow time calculated.")
df[["request_id", "waiting_hours", "processing_hours", "total_time_hours"]].head()
```

```python
df["sla_violated"] = (
    df["total_time_hours"] > df["sla_hours"]
)

print("SLA violations detected:")
print(df["sla_violated"].value_counts())
```

```python
df["delay_hours"] = (
    df["total_time_hours"] - df["sla_hours"]
).clip(lower=0)

print("Delay calculation completed.")

df[[
    "request_id",
    "total_time_hours",
    "sla_hours",
    "delay_hours"
]].head()
```

```python
df["risk_level"] = np.where(
    df["delay_hours"] > 48,
    "Critical",
    np.where(
        df["delay_hours"] > 24,
        "High",
        np.where(
            df["delay_hours"] > 0,
            "Medium",
            "Low"
            
        )
    )
)

print("Risk levels:")
print(df["risk_level"].value_counts())
```

```python
delayed_count = len(df[df["status"] == "Delayed"])

print("Total delayed requests:", delayed_count)
```

```python
sla_breaches = 0

for i in range(len(df)):
    if df.iloc[i]["total_time_hours"] > df.iloc[i]["sla_hours"]:
        sla_breaches = sla_breaches + 1

print("Total SLA breaches:", sla_breaches)
```

```python
total_requests = len(df)

sla_compliance = 100 - (sla_breaches / total_requests * 100)

print("SLA Compliance:", round(sla_compliance, 2), "%")
```

```python
completed_count = len(df[df["status"] == "Completed"])
total_requests = len(df)

completion_rate = completed_count / total_requests * 100

print("Completed requests:", completed_count)
print("Completion Rate:", round(completion_rate, 2), "%")
```

```python
print("Most delayed request:")

highest_delay = -1
highest_request = ""

for i in range(len(df)):
    delay = df.iloc[i]["delay_hours"]
    
    if delay > highest_delay:
        highest_delay = delay
        highest_request = df.iloc[i]["request_id"]

print("Request ID:", highest_request)
print("Delay:", round(highest_delay, 2), "hours")
```

```python
print("Process bottleneck analysis:")

highest_average = -1
worst_process = ""

for process in df["process"].unique():
    values = df[df["process"] == process]["delay_hours"].tolist()
    
    total = 0
    for value in values:
        total = total + value
    
    average = total / len(values)
    
    print(process, ":", round(average, 2), "hours")
    
    if average > highest_average:
        highest_average = average
        worst_process = process

print()
print("Main process bottleneck:", worst_process)
print("Average delay:", round(highest_average, 2), "hours")
```

```python
print("Department bottleneck analysis:")

highest_average = -1
worst_department = ""

for department in df["department"].unique():
    values = df[df["department"] == department]["delay_hours"].tolist()
    
    total = 0
    for value in values:
        total = total + value
    
    average = total / len(values)
    
    print(department, ":", round(average, 2), "hours")
    
    if average > highest_average:
        highest_average = average
        worst_department = department

print()
print("Main department bottleneck:", worst_department)
print("Average delay:", round(highest_average, 2), "hours")
```

```python
print("Priority risk analysis:")

highest_average = -1
highest_risk_priority = ""

for priority in df["priority"].unique():
    values = df[df["priority"] == priority]["delay_hours"].tolist()
    
    total = 0
    for value in values:
        total = total + value
    
    average = total / len(values)
    
    print(priority, ":", round(average, 2), "hours")
    
    if average > highest_average:
        highest_average = average
        highest_risk_priority = priority

print()
print("Highest-risk priority:", highest_risk_priority)
print("Average delay:", round(highest_average, 2), "hours")
```

```python
print("Workflow risk summary:")

delayed_requests = 0
sla_breached_requests = 0

for i in range(len(df)):
    if df.iloc[i]["delay_hours"] > 0:
        delayed_requests = delayed_requests + 1
    
    if df.iloc[i]["total_time_hours"] > df.iloc[i]["sla_hours"]:
        sla_breached_requests = sla_breached_requests + 1

print("Delayed requests:", delayed_requests)
print("SLA breached requests:", sla_breached_requests)
print("Total requests:", len(df))
```

```python
# Step 32: Calculate workflow risk score

df["risk_score"] = 0

for i in range(len(df)):
    score = 0
    
    if df.iloc[i]["delay_hours"] > 0:
        score = score + 30
    
    if df.iloc[i]["delay_hours"] > 24:
        score = score + 20
    
    if df.iloc[i]["delay_hours"] > 48:
        score = score + 20
    
    if df.iloc[i]["rework_count"] > 0:
        score = score + 15
    
    if df.iloc[i]["priority"] == "Critical":
        score = score + 15
    
    df.loc[df.index[i], "risk_score"] = score

print("Risk score created.")
print(df[["request_id", "delay_hours", "rework_count", "risk_score"]].head(10))
```

```python

df["risk_category"] = "Low"

for i in range(len(df)):
    score = df.iloc[i]["risk_score"]
    
    if score >= 70:
        df.loc[df.index[i], "risk_category"] = "Critical"
    elif score >= 50:
        df.loc[df.index[i], "risk_category"] = "High"
    elif score >= 30:
        df.loc[df.index[i], "risk_category"] = "Medium"

print("Risk categories created.")
print(df[["request_id", "risk_score", "risk_category"]].head(10))
```
